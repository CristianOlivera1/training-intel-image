```python
import tensorflow as tf

print("Verificando dispositivos...")
dispositivos_gpu = tf.config.list_physical_devices('GPU')

if dispositivos_gpu:
    print(f"¡Éxito! Se detectaron {len(dispositivos_gpu)} GPU(s).")
    for gpu in dispositivos_gpu:
        print("Detalles:", gpu)
else:
    print("Sigue sin detectarse la GPU.")
```
